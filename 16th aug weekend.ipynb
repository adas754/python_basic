```python
# if the number is divisible by 5 and 7 both then print ok
# if the number is divisible by 5 but not 7 then print hi
#if the number is divisible by 7 but not 5 then print hlw
# if not divisible by either 5 or 7 then print not a good number
n=int(input('enter number'))
if n%5==0 and n%7==0:
    print('ok')
elif n%5==0 and n%7!=0:
    print('hi')
elif n%7==0 and n%5!=0:
    print('hlw')
elif n%7!=0 and n%5!=0:
    print('not a good number')
```

```python
# 1 to 10
# 8 lucky no
# 5 
# choose the higher
# 10
# choose the lower 
# u will choose the correct number it will u won the game
# your attempt is times 
import random
jackpot=random.randint(1,10)
user_number=int(input('choose the number'))
count=1
while jackpot!=user_number:
    if user_number<jackpot:
        print('choose the higher')
    else:
        print('choose the lower')
    user_number=int(input('choose the number'))
    count+=1
else:
    print('you won the game')
    print('your attempt',count,'times')
```
